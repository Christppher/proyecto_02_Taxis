## Preparación

```python
# importamos librerias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
# cargarmos los datos con las fechas como indices
data = pd.read_csv('taxi.csv', index_col=[0], parse_dates=[0])
```

```python
# organizamos los inices y ordenamos por hora
data.sort_index(inplace=True)
data_resampled = data.resample('1h').sum()
```

## Análisis

```python
data_decompose = seasonal_decompose(data_resampled['num_orders'], period=24)
data_decompose.plot()
```

```python
# Agrupamos por la hora del indice (0-23) y sacamos la media
daily_pattern = data_resampled.groupby(data_resampled.index.hour)['num_orders'].mean()

# Graficamos el patrón
plt.figure(figsize=(12, 6))
daily_pattern.plot(kind='bar')
plt.title('Patron Diario Promedio de Pedidos de Taxi')
plt.xlabel('Hora del Dia (0-23)')
plt.ylabel('Pedidos Promedio')
plt.grid(axis='y')
plt.show()
```

## Formación

```python
# creamos una funcion
def make_features(data, max_lag, rolling_mean_size):
    df = data.copy()
    # caracteristicas de calendario
    df['hour'] = df.index.hour
    df['dayofweek'] = df.index.dayofweek
    df['month'] = df.index.month
    df['day'] = df.index.day
    # caracteristicas de desfase
    for lag in range(1, max_lag + 1):
        df[f'lag_{lag}'] = df['num_orders'].shift(lag)
    return df
```

```python
data_features = make_features(data_resampled, 24, 6)
data_features.head()
```

```python
# eliminamos NaN
data_features = data_features.dropna()
```

```python
# preparamos los datos para entrenar al modelo
# dividimos los datos y creamos los features
train, test = train_test_split(data_features, shuffle=False, test_size=0.1)
x_train = train.drop(['num_orders'], axis=1)
y_train = train['num_orders']
x_test = test.drop(['num_orders'], axis=1)
y_test = test['num_orders']
```

```python
# Modelo de Regresion Lineal
model_lr = LinearRegression()
model_lr.fit(x_train, y_train)
predictions_lr = model_lr.predict(x_test)
mse_lr = mean_squared_error(y_test, predictions_lr)
# sacamos el RMSE
recm_lr = np.sqrt(mse_lr)
print(f"\nEl RMSE de la Regresion Lineal es: {recm_lr:.2f}")
```

```python
# Modelo RandomForest
model_rf = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=12345)
model_rf.fit(x_train, y_train)
predictions_rf = model_rf.predict(x_test)
mse_rf = mean_squared_error(y_test, predictions_rf)
recm_rf = np.sqrt(mse_rf)
print(f"RMSE de Random Forest: {recm_rf:.2f}")
```
